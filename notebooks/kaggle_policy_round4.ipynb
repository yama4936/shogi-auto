自分のv3重みから追加学習します。モデルや入力データをこの公開Notebookへ埋め込んでいません。
無料GPU残量をUIで確認し、REMAINING_FREE_GPU_HOURSへ入力。既存PyTorch/CUDAを更新しません。
最大40epoch・validation早期終了。testは計算せず、全3回の方法確定後にCPUで最終検証します。
学習累計1200秒・全体1500秒＋終了猶予15秒。保存後すぐセッションOFF。対局はCPUで別実行。


```python
SOURCE_BASE64=''
SOURCE_ZIP_SHA256='d9aed9099766085bed286a951482a15e70766fb0c99a0d8ab3c36c6375231433'
PINNED_CODE_COMMIT='eaa315becdc9eb041ebe8d98ac6633c167151ca0'
DATA_ZIP_SHA256='7a37d2ff044a37a621f97a05737e427275f3479c09fc9edf13b3f87b4f0cbe2e'
DATA_BASE64=None
import base64, hashlib, json, os, pathlib, subprocess, sys, time, zipfile, io
started=time.monotonic()
# Fill this only after checking the current Kaggle free quota in the UI.
REMAINING_FREE_GPU_HOURS=None
os.environ['CUBLAS_WORKSPACE_CONFIG']=':4096:8'
import torch
if REMAINING_FREE_GPU_HOURS is None:raise RuntimeError('Check free GPU quota before running')
if not torch.cuda.is_available():raise RuntimeError('CUDA is required; no CPU fallback')
visible=torch.cuda.device_count()
if visible not in (1,2):raise RuntimeError('Reviewed first experiment allows at most two allocated GPUs')
if REMAINING_FREE_GPU_HOURS < visible*1520/3600+.05:raise RuntimeError('Insufficient confirmed free GPU allowance')
if str(torch.__version__).split('+')[0]!='2.11.0':raise RuntimeError('PyTorch differs from the reviewed environment; do not auto-upgrade')
raw=base64.b64decode(SOURCE_BASE64,validate=True)
assert hashlib.sha256(raw).hexdigest()==SOURCE_ZIP_SHA256
code_dir=pathlib.Path('/kaggle/working/shogi-policy-round4-code')
code_dir.mkdir(exist_ok=True)
with zipfile.ZipFile(io.BytesIO(raw)) as z:
    for entry in z.infolist():
        p=pathlib.PurePosixPath(entry.filename)
        if p.is_absolute() or '..' in p.parts or entry.file_size>2_000_000:raise ValueError('Invalid source bundle path/size')
    z.extractall(code_dir)
origin=json.loads((code_dir/'SOURCE_POLICY.json').read_text())
assert origin['commit']==PINNED_CODE_COMMIT
INPUT_CHECKSUMS={'data.npz': 'b1efbe76e8c695875ff44861ddf5d6f7743ebe31841c3390b308705035bccbc5', 'data-manifest.json': '0dbee218916eeb9987cc32ac40cea689d50bd3b78a8ed70f2ce0ef1e1199ed28', 'start-policy.json': 'd412395e58bb0e1dae562c506d44beb2f755e065c6a92494c3edcbba64aacf0a'}
archives=list(pathlib.Path('/kaggle/input').glob('**/policy-round4-inputs.zip'))
archives+=list(pathlib.Path('/kaggle/input').glob('**/policy-round4-inputs.bundle'))
allowed={'data.npz','data-manifest.json','start-policy.json','SHA256SUMS.json'}
if len(archives)==1:
    data_raw=archives[0].read_bytes()
    if hashlib.sha256(data_raw).hexdigest()!=DATA_ZIP_SHA256:raise ValueError('Input ZIP checksum mismatch')
    with zipfile.ZipFile(io.BytesIO(data_raw)) as z:
        if set(z.namelist())!=allowed or sum(v.file_size for v in z.infolist())>200_000_000:
            raise ValueError('Unexpected round input archive')
        items={name:z.read(name) for name in allowed}
    input_zip_verified=True
elif not archives:
    # Kaggle may expand uploaded ZIPs into individual dataset files.
    directories=[p.parent for p in pathlib.Path('/kaggle/input').glob('**/data-manifest.json')
                 if all((p.parent/name).is_file() for name in allowed)]
    if len(directories)!=1:raise RuntimeError('Attach exactly one private round input dataset')
    items={name:(directories[0]/name).read_bytes() for name in allowed}
    input_zip_verified=False
else:raise RuntimeError('Duplicate round input archives')
if json.loads(items['SHA256SUMS.json'])!=INPUT_CHECKSUMS:raise ValueError('Input checksum manifest differs')
data_dir=pathlib.Path('/kaggle/working/shogi-policy-round4-data');data_dir.mkdir(exist_ok=True)
for name,expected in INPUT_CHECKSUMS.items():
    if hashlib.sha256(items[name]).hexdigest()!=expected:raise ValueError('Input item checksum mismatch')
    (data_dir/name).write_bytes(items[name])
manifest=json.loads(items['data-manifest.json'])
if manifest['round_id']!=4:raise ValueError('Wrong round input')
if manifest['warm_start_file_sha256']!=INPUT_CHECKSUMS['start-policy.json']:raise ValueError('Wrong starting model')
env=os.environ.copy();env['PYTHONPATH']=str(code_dir/'src')+os.pathsep+str(code_dir/'vendor')
env['OMP_NUM_THREADS']='1';env['MKL_NUM_THREADS']='1'
destination=pathlib.Path('/kaggle/working/shogi-policy-round4');destination.mkdir(exist_ok=True)
command=[sys.executable,'-m','shogi_auto.policy.train','--data-dir',str(data_dir),'--run-dir',str(destination),
         '--device','cuda','--epochs','40','--batch-size','128','--max-wall-seconds','1200',
         '--initial-model',str(data_dir/'start-policy.json'),'--learning-rate','0.0003',
         '--weight-decay','0.03','--label-smoothing','0.05','--patience','6','--min-epochs','6','--seed','314162','--value-weight','2','--balanced-sample-weight','2','--defer-test']
log=destination/'training-process.log';timed_out=False
with log.open('a') as stream:
    process=subprocess.Popen(command,env=env,stdout=stream,stderr=subprocess.STDOUT)
    try:process.wait(timeout=max(1,1500-(time.monotonic()-started)))
    except subprocess.TimeoutExpired:
        timed_out=True;process.terminate()
        try:process.wait(timeout=15)
        except subprocess.TimeoutExpired:process.kill();process.wait()
report_path=destination/'training-report.json'
training=json.loads(report_path.read_text()) if report_path.exists() else {}
result={'round_id':4,'status':training.get('status','failed'),'exit_code':process.returncode,'watchdog_triggered':timed_out,
        'wall_seconds':time.monotonic()-started,'source':origin,'source_zip_sha256':SOURCE_ZIP_SHA256,
        'data_zip_sha256':DATA_ZIP_SHA256,'visible_gpus':visible,'gpu_names':[torch.cuda.get_device_name(i) for i in range(visible)],
        'gpu_training_verified':bool(training.get('gpu_training_verified',False)),
        'cuda_optimizer_steps_total':training.get('cuda_optimizer_steps_total',0),
        'strength_improvement_verified':False,'cpu_arena_pending':True,'input_zip_sha256_verified':input_zip_verified,'input_payload_sha256_verified':True,
        'note':'Save private outputs and switch the session OFF immediately; this Notebook does no CPU arena or teacher generation.'}
if process.returncode!=0:result['status']='failed'
result['estimated_allocation_gpu_hours']=result['wall_seconds']*visible/3600
(destination/'run-report.json').write_text(json.dumps(result,ensure_ascii=False,indent=2)+'\n')
print(json.dumps(result,ensure_ascii=False,indent=2))
print(log.read_text()[-5000:])
# Text output can be read through Library's native text route if signed binary
# transfers are unavailable in the separate CPU environment.
names=['initial-policy.json','best-policy.json','training-report.json','training-settings.json',
       'run-report.json','optimizer-accounting.json','training-process.log']
memory=io.BytesIO()
with zipfile.ZipFile(memory,'w',zipfile.ZIP_DEFLATED) as z:
    checks={}
    for name in names:
        path=destination/name
        if path.exists():
            payload=path.read_bytes();checks[name]=hashlib.sha256(payload).hexdigest();z.writestr(name,payload)
    z.writestr('SHA256SUMS.json',json.dumps(checks,ensure_ascii=False,indent=2)+'\n')
payload=memory.getvalue();encoded=base64.b64encode(payload).decode()
header={'format':'shogi-policy-v2-results-base64zip','zip_bytes':len(payload),
        'zip_sha256':hashlib.sha256(payload).hexdigest(),'data_zip_sha256':DATA_ZIP_SHA256,
        'source_commit':PINNED_CODE_COMMIT}
text=json.dumps(header,ensure_ascii=False)+'\nBEGIN_BASE64\n'+'\n'.join(encoded[i:i+256] for i in range(0,len(encoded),256))+'\nEND_BASE64\n'
(destination/'policy-round4-return.txt').write_text(text)
print('CPU return file:',str(destination/'policy-round4-return.txt'))
print(json.dumps(header,ensure_ascii=False))
```
